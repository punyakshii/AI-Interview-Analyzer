```python
skills_db = [

    "python",
    "machine learning",
    "deep learning",
    "nlp",
    "sql",
    "tensorflow",
    "pandas",
    "data science",
    "java",
    "c++"
]

resume_lower = resume_text.lower()

detected_skills = []

for skill in skills_db:

    if skill in resume_lower:

        detected_skills.append(skill)

print("Detected Skills:")
print(detected_skills)
```

```python
sentiment_model = pipeline(
    "sentiment-analysis"
)
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

class FlanT5Generator:
    def __init__(self, model_name="google/flan-t5-base"):
        self.tokenizer = AutoTokenizer.from_pretrained(model_name)
        self.model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

    def __call__(self, prompt_text, max_new_tokens=120, num_return_sequences=1):
        # For T5, the prompt is typically used as the input
        inputs = self.tokenizer(prompt_text, return_tensors="pt")
        outputs = self.model.generate(
            inputs["input_ids"],
            attention_mask=inputs["attention_mask"],
            max_new_tokens=max_new_tokens,
            num_return_sequences=num_return_sequences
        )
        decoded_outputs = [self.tokenizer.decode(output, skip_special_tokens=True) for output in outputs]
        # Mimic the pipeline output format: list of dicts with 'generated_text' key
        return [{'generated_text': text} for text in decoded_outputs]

# Instantiate the custom generator
question_generator = FlanT5Generator(model_name="google/flan-t5-base")
```

```python
prompt = f"""
Generate 5 professional interview questions for a candidate with these skills:

{detected_skills}

Return only questions.
"""

result = question_generator(
    prompt,
    max_new_tokens=256
)

print(result[0]['generated_text'])
```

```python
tool = language_tool_python.LanguageTool('en-US')
```

```python
confidence_words = [

    "implemented",
    "developed",
    "created",
    "designed",
    "led",
    "confident",
    "successfully",
    "optimized",
    "improved"
]
```

```python
filler_words = [

    "umm",
    "uh",
    "like",
    "actually",
    "basically"
]
```

```python
import pydub
from google.colab import files
import speech_recognition as sr

questions = [
    "Tell me about yourself.",
    "Explain a machine learning project you worked on.",
    "What is NLP and where is it used?",
    "What challenges did you face in your project?",
    "Why should we hire you?"
]

answers = []

for q in questions:

    print("\n")
    print("QUESTION:")
    print(q)

    print("\nChoose Answer Method")
    print("1. Text")
    print("2. Speech")

    choice = input("Enter choice (1 or 2): ")

    # --------------------------------
    # TEXT INPUT
    # --------------------------------

    if choice == "1":

        ans = input("Type your answer: ")

        answers.append(ans)

    # --------------------------------
    # SPEECH INPUT
    # --------------------------------

    elif choice == "2":

        print("\nUpload WAV audio file")

        uploaded = files.upload()

        if uploaded: # Check if any file was actually uploaded

            recognizer = sr.Recognizer()

            original_file = list(uploaded.keys())[0]

            # Convert to WAV if not already WAV
            if not original_file.lower().endswith('.wav'):
                print(f"Converting {original_file} to WAV...")
                audio_segment = pydub.AudioSegment.from_file(original_file)
                wav_file_name = original_file.rsplit('.', 1)[0] + '.wav'
                audio_segment.export(wav_file_name, format="wav")
                audio_file = wav_file_name
            else:
                audio_file = original_file

            try:
                with sr.AudioFile(audio_file) as source:

                    audio = recognizer.record(source)

                speech_text = recognizer.recognize_google(audio)

                print("\nConverted Speech to Text:")

                print(speech_text)

                answers.append(speech_text)

            except Exception as e:

                print(f"Speech Recognition Failed: {e}")

                answers.append("")
        else:
            print("No file uploaded. Skipping speech input for this question.")
            answers.append("") # Append an empty string if no file was uploaded

    else:

        print("Invalid Choice")

        answers.append("")
```

```python
results = []

for i in range(len(answers)):

    answer = answers[i]

    # --------------------------------
    # SENTIMENT ANALYSIS
    # --------------------------------

    sentiment = sentiment_model(answer)

    sentiment_label = sentiment[0]['label']

    sentiment_score = sentiment[0]['score']


    # --------------------------------
    # GRAMMAR ANALYSIS
    # --------------------------------

    grammar_matches = tool.check(answer)

    grammar_errors = len(grammar_matches)

    grammar_feedback = []

    for match in grammar_matches:

        feedback = {

            "mistake": match.context,

            "message": match.message,

            "suggestions": match.replacements[:3]
        }

        grammar_feedback.append(feedback)


    # --------------------------------
    # TECHNICAL ANALYSIS
    # --------------------------------

    technical_score = 0

    detected_keywords = []

    for skill in detected_skills:

        if skill.lower() in answer.lower():

            technical_score += 1

            detected_keywords.append(skill)


    # --------------------------------
    # CONFIDENCE ANALYSIS
    # --------------------------------

    confidence_score = 0

    confidence_detected = []

    for word in confidence_words:

        if word in answer.lower():

            confidence_score += 1

            confidence_detected.append(word)


    # --------------------------------
    # FILLER WORD ANALYSIS
    # --------------------------------

    filler_count = 0

    for filler in filler_words:

        filler_count += answer.lower().count(filler)


    # --------------------------------
    # ANSWER LENGTH
    # --------------------------------

    word_count = len(answer.split())


    # --------------------------------
    # COMMUNICATION SCORE
    # --------------------------------

    communication_score = (

        technical_score * 2

        + confidence_score * 1.5

        + sentiment_score * 2

        - grammar_errors

        - filler_count
    )


    # --------------------------------
    # IMPROVEMENT SUGGESTIONS
    # --------------------------------

    improvements = []

    if grammar_errors > 2:

        improvements.append(
            "Improve grammar and sentence structure."
        )

    if technical_score < 2:

        improvements.append(
            "Add more technical explanation."
        )

    if confidence_score < 2:

        improvements.append(
            "Use stronger confident words."
        )

    if filler_count > 2:

        improvements.append(
            "Reduce filler words."
        )

    if word_count < 20:

        improvements.append(
            "Give more detailed answers."
        )


    # --------------------------------
    # STORE RESULT
    # --------------------------------

    results.append({

        "question": questions[i],

        "answer": answer,

        "sentiment_label": sentiment_label,

        "sentiment_score": round(
            sentiment_score,
            2
        ),

        "grammar_errors": grammar_errors,

        "grammar_feedback": grammar_feedback,

        "technical_score": technical_score,

        "technical_keywords": detected_keywords,

        "confidence_score": confidence_score,

        "confidence_words": confidence_detected,

        "filler_count": filler_count,

        "word_count": word_count,

        "communication_score": round(
            communication_score,
            2
        ),

        "improvements": improvements
    })
```

```python
total_score = 0

print("\n")
print("========== AI INTERVIEW REPORT ==========")

for r in results:

    print("\n")
    print("QUESTION:")
    print(r["question"])

    print("\nANSWER:")
    print(r["answer"])

    print("\n----------------------------------")


    # SENTIMENT

    print("SENTIMENT ANALYSIS")

    print(
        "Label:",
        r["sentiment_label"]
    )

    print(
        "Score:",
        r["sentiment_score"]
    )

    print(
        "Reason: Tone and positivity detected."
    )


    # GRAMMAR

    print("\nGRAMMAR ANALYSIS")

    print(
        "Grammar Errors:",
        r["grammar_errors"]
    )

    if len(r["grammar_feedback"]) > 0:

        for g in r["grammar_feedback"]:

            print("\nMistake:")

            print(g["mistake"])

            print("Explanation:")

            print(g["message"])

            print("Suggestions:")

            print(g["suggestions"])


    # TECHNICAL

    print("\nTECHNICAL ANALYSIS")

    print(
        "Technical Score:",
        r["technical_score"]
    )

    print(
        "Detected Keywords:",
        r["technical_keywords"]
    )


    # CONFIDENCE

    print("\nCONFIDENCE ANALYSIS")

    print(
        "Confidence Score:",
        r["confidence_score"]
    )

    print(
        "Detected Confidence Words:",
        r["confidence_words"]
    )


    # COMMUNICATION

    print("\nCOMMUNICATION ANALYSIS")

    print(
        "Communication Score:",
        r["communication_score"]
    )

    print(
        "Filler Words:",
        r["filler_count"]
    )

    print(
        "Answer Length:",
        r["word_count"],
        "words"
    )


    # IMPROVEMENTS

    print("\nIMPROVEMENT SUGGESTIONS")

    for imp in r["improvements"]:

        print("-", imp)

    print("\n===================================")

    total_score += r["communication_score"]
```

```python
overall_score = total_score / len(results)

print("\n")
print("========== FINAL RESULT ==========")

print(
    "Overall Interview Score:",
    round(overall_score, 2)
)

if overall_score >= 8:

    print("Performance: EXCELLENT")

elif overall_score >= 5:

    print("Performance: GOOD")

else:

    print("Performance: NEEDS IMPROVEMENT")
```
